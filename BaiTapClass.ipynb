### Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ

## Đầu vào
### Đầu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.

## Đầu ra
### In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14). (Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 


```python
class Cylinder:
    def __init__(self, ban_kinh:float, chieu_cao:float):
        self.r = ban_kinh
        self.h = chieu_cao
    def dien_tich(self):
        A = 2 * 3.14 * (self.r ** 2) + 2 * 3.14 * self.r * self.h
        return "%.2f"%A
    def the_tich(self):
        V = 3.14 * (self.r ** 2) * self.h
        return "%.2f"%V
ban_kinh, chieu_cao = map(float, input().split())
thong_so = Cylinder(ban_kinh, chieu_cao)
print(thong_so.dien_tich())
print(thong_so.the_tich())
```

## Đầu vào:

### Nhập vào tọa độ một điểm M bất kỳ

## Đầu ra:

### Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ số phần thập phân) 

```python
class Diem:
    def __init__(self, x:float, y:float):
        self.x = x
        self.y = y
    def kiem_tra(self):
        Ox = False
        Oy = False
        if self.x == 0 and self.y != 0:
            Oy = True
        if self.x != 0 and self.y == 0:
            Ox = True
        if self.x == 0 and self.y == 0:
            Ox, Oy = True, True
        if Ox and not Oy:
            return f"Thuộc Ox, khoảng cách: {self.x}"
        elif not Ox and Oy:
            return f"Thuộc Oy, khoảng cách: {self.y}"
        elif Ox and Oy:
            return f"Là gốc tọa độ, khoảng cách: 0"
        else:
            A = ((self.x ** 2) + (self.y ** 2)) ** 0.5
            return f"Không thuộc Ox và Oy, khoảng cách: {"%.2f"%A}"
x, y = map(float, input().split())
toa_do = Diem(x, y)
print(toa_do.kiem_tra())
```

### Xây dựng lớp Calculator (máy tính bỏ túi) với các phương thức để thực hiện các phép toán 1. add (cộng), 2. subtract(trừ), 3.product (nhân) và 4. divide(chia), 5. power(hàm mũ), 6. mod (lấy phần dư) và phương thức 7. set_numbers(cập nhật 2 số mới) nếu người dùng muốn tính toán trên 2 số khác.

### Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương thức thực hiện một phép tính và trả về kết quả tương ứng

```python
class Calculator:
    def __init__(self, number1: float, number2: float):
        self.x = number1
        self.y = number2
    def add(self) -> float:
        return self.x + self.y
    def subtract(self) -> float:
        return self.x - self.y
    def product(self) -> float:
        return self.x * self.y
    def devide(self) -> float:
        return self.x / self.y
    def power(self) -> float:
        return self.x ** self.y
    def mod(self) -> float:
        return self.x % self.y
    def set_number(self, newnum1: float, newnum2: float):
        self.x = newnum1
        self.y = newnum2
print(f"Chào mừng đến với máy tính bỏ túi")
x, y = map(float, input().split())
my_cal = Calculator(x, y)
while True:
    print(f"1. Cộng")
    print(f"2. Trừ")
    print(f"3. Nhân")
    print(f"4. Chia")
    print(f"5. Hàm mũ")
    print(f"6. Lấy dư")
    print(f"7. Cập nhật 2 số mới")
    choice = int(input())
    if choice == 1:
        print(f"Kết quả của phép cộng: {x} + {y} = {my_cal.add()}")
    elif choice == 2:
        print(f"Kết quả của phép trừ: {x} - {y} = {my_cal.subtract()}")
    elif choice == 3:
        print(f"Kết quả của phép nhân: {x} x {y} = {my_cal.product()}")
    elif choice == 4:
        print(f"Kết quả của phép chia: {x} / {y} = {my_cal.devide()}")
    elif choice == 5:
        print(f"Kết quả của hàm mũ: {x} ^ {y} = {my_cal.power()}")
    elif choice == 6:
        print(f"Kết quả của số dư khi lấy {x} / {y}: {my_cal.mod()}")
    elif choice == 7:
        a, b = map(int, input().split())
        my_cal.set_number(a, b)
        print(f"Đã cập nhật")
    else:
        print(f"Lựa chọn không lợp lệ")
    OUT = input()
    if OUT.lower() == "yes":
        print(f"Cảm ơn và tạm biệt !!!")
        break
```

## [ShoppingCart]

### Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi sản phẩm là một từ điển có name (tên) và price(giá).

### Với các phương thức sau:

### 1. Khởi tạo với danh sách rỗng
### 2. Thêm vào giỏ hàng một sản phẩm mới
### 3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
### 4. Kiểm tra xem giỏ hàng có rỗng không
### 5. Tính tổng tiền giỏ hàng
### 6. Hiển thị toàn bộ sản phẩm trong giỏ
### 7. Xóa toàn bộ giỏ hàng

### Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7. Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có muốn thoát khỏi chương trình không?" 


```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add(self, item: str, price: float):
        new_item = {"name": item, "price": price}
        self.items.append(new_item)
    def removE(self, name: str):
        for i in self.items:
            if i["name"].lower() == name.lower():
                self.items.remove(i)
                break
    def empty(self) -> bool:
        return len(self.items) == 0
    def total_money(self) -> float:
        tong_tien = 0
        for i in self.items:
            tong_tien += i["price"]
        return tong_tien
    def display(self):
        lisT = []
        for i in self.items:
            lisT.append(i["name"])
        if len(lisT) == 0:
            return f"empty"
        else:
            return lisT
    def remove_all(self):
        self.items.clear()
Things = ShoppingCart()
while True:
    print("\n================ MENU ================")
    print("1. Thêm sản phẩm vào giỏ")
    print("2. Xóa 1 sản phẩm khỏi giỏ (theo tên)")
    print("3. Kiểm tra giỏ hàng có rỗng không?")
    print("4. Tính tổng tiền thanh toán")
    print("5. Hiển thị toàn bộ giỏ hàng")
    print("6. Xóa toàn bộ giỏ hàng")
    print("======================================")
    choice = int(input())
    if choice == 1:
        ten = input()
        gia = float(input())
        Things.add(ten, gia)
    elif choice == 2:
        ten = input()
        Things.removE(ten)
    elif choice == 3:
        if Things.empty():
            print(f"Giỏ hàng rỗng")
        else:
            print(f"Giỏ hàng ko rỗng")
    elif choice == 4:
        print(f"Tổng số tiền là {Things.total_money()}")
    elif choice == 5:
        print(f"Giỏ hàng hiện đang có: {Things.display()}")
    elif choice == 6:
        Things.remove_all()
    else:
        print(f"Lựa chọn không hợp lệ")
    out = input()
    if out.lower() == "yes":
        print(f"Cảm ơn đã lựa chọn dịch vụ")
        break
```

### Xây dựng lớp Rectangle(w, l) với các phương thức:

### Area() trả về diện tích.

### Perimeter() trả về chu vi.

### Scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), không trả về gì.

## Input

### w l k (số thực dương)

## Output

### area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân). 

```python
class Rectangle:
    def __init__(self, width: float, length: float):
        self.w = width
        self.l = length
    def area(self) -> float:
        A = self.w * self.l
        return "%.2f"%A
    def preimeter(self) -> float:
        P = 2*(self.w + self.l)
        return "%.2f"%P 
    def scale(self, k:int) -> None:
        self.w *= k
        self.l *= k
width, length, k = map(float, input().split())
stat = Rectangle(width, length)
stat.scale(k)
print(stat.area())
print(stat.preimeter())
```

## Yêu cầu

### Lớp BankAccount(owner, balance) với: deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.

### __str__() in owner: balance.

## Input

### Dòng 1: owner initial_balance

### Dòng 2: n (số lệnh)

### Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.

## Output

### Số dư cuối cùng (in số thực, 2 chữ số). 


```python
class BankAccount:
    def __init__(self, owner:str, balance:int):
        self.owner = owner
        self.balance = balance
    def deposit(self, x:int):
        self.balance += x
    def withdraw(self, x:int):
        if x <= self.balance:
            self.balance -= x
    def __str__(self):
        return f"{self.owner}: {self.balance}"
```

## Yêu cầu

### Lớp Student(name) với:

### add_score(subject, score)

### avg() trả về điểm trung bình.

### rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor".

## Input

### Dòng 1: name
### Dòng 2: n
### n dòng tiếp: subject score (score là số thực 0–10)

## Output

### name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self, name:str):
        self.name = name
        self.score = []
    def add_score(self, subject: str, score: float):
        self.score.append(score)
    def avg(self):
        tong = 0
        for i in self.score:
            tong += i
        AVG = tong/len(self.score)
        return AVG
    def rank(self):
        diemtb = self.avg()
        if diemtb >= 8:
            return f"Excellent"
        elif 6 <= diemtb < 8:
            return f"Good"
        elif 5 <= diemtb < 6:
            return f"Average"
        else:
            return f"Poor"
name = input()
ho_so = Student(name)
n = int(input())
for i in range(n):
    string = input().split()
    subject = string[0]
    score = float(string[1])
    ho_so.add_score(subject, score)
print(ho_so.rank())
```

### In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập
phân.

```python
import math
class Radian:
    def __init__(self, x1, y1, z1, x2, y2, z2, x3, y3, z3, x4, y4, z4):
        self.A = [x1, y1, z1]
        self.B = [x2, y2, z2]
        self.C = [x3, y3, z3]
        self.D = [x4, y4, z4]
    def Phi(self):
        BA = [self.A[0] - self.B[0], self.A[1] - self.B[1], self.A[2] - self.B[2]]
        BC = [self.C[0] - self.B[0], self.C[1] - self.B[1], self.C[2] - self.B[2]]
        BD = [self.D[0] - self.B[0], self.D[1] - self.B[1], self.D[2] - self.B[2]]
        N_BAC = [BA[1] * BC[2] - BC[1] * BA[2], BA[2] * BC[0] - BA[0] * BC[2], BA[0] * BC[1] - BC[0] * BA[1]]
        N_BCD = [BD[1] * BC[2] - BC[1] * BD[2], BD[2] * BC[0] - BC[2] * BD[0], BD[0] * BC[1] - BC[0] * BD[1]]
        tich_cham = 0
        for i in range(3):
            tich_cham += N_BAC[i]*N_BCD[i]
        chuan_bac = ((N_BAC[0])**2 + (N_BAC[1])**2 + (N_BAC[2])**2)**0.5
        chuan_bcd = ((N_BCD[0])**2 + (N_BCD[1])**2 + (N_BCD[2])**2)**0.5
        theta = math.acos((tich_cham)/(chuan_bac * chuan_bcd))
        phi = (theta/(2*3.14))*360
        return f"{"%.2f"%phi}"
x1, y1, z1, x2, y2, z2, x3, y3, z3, x4, y4, z4 = map(float, input().split())
thong_so = Radian(x1, y1, z1, x2, y2, z2, x3, y3, z3, x4, y4, z4)
print(thong_so.Phi())
```
