W5A16: Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def Amstronhg(a):
    tong = 0
    for i in a:
        tong += int(i)**(len(str(a)))
    if tong == int(a):
        return True
    else:
        return False
x = input()
print(Amstronhg(x))
```

W5A22: Viết hàm sum_halves(s) nhận chuỗi 6 chữ số và trả về tổng của hai số 3 chữ số đầu và 3 chữ số cuối.

```python
def sum_halve(s):
    if len(str(s)) == 6:
        tong1 = 0
        tong2 = 0
        for i in str(s):
            if 0 <= int(i) <= 2:
                tong1 += int(i)
            if 3 <= int(i) <= 5:
                tong2 += int(i)
        return f'{tong1}, {tong2}'
    if len(str(s)) != 6:
        return f'invalid number'
n = int(input())
print(sum_halve(n))
```

W5A23: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    delta = b**2 - 4*a*c
    if delta < 0:
        return f'Vô nghiệm'
    if delta == 0:
        x = -b/(2*a)
        return f'Phương trình có 1 nghiệm kép là {x}'
    if delta > 0:
        x1 = (-b + (delta ** 0.5))/(2*a)
        x2 = (-b - (delta ** 0.5))/(2*a)
        return f'Phương trình có 2 nghiệm phân biệt lần lượt là {x1} và {x2}'
a, b, c = map(int, input().split())
print(quadratic_roots_count(a, b, c))
```

W5A24: Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
    if x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2) == 0: #Thẳng hàng -> Diện tích tam giác = 0
        return True
    else:
        return False
#True: Thẳng hàng
#False: Không thẳng hàng
x1,y1,x2,y2,x3,y3 = map(int, input().split())
print(are_collinear(x1,y1,x2,y2,x3,y3))
```

W5A28: Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math
def angle(x1, y1, x2, y2, x3, y3):
    A = (x1-x2)*(x3-x2)+(y1-y2)*(y3-y2)
    BA = ((x2-x1)**2 + (y2-y1)**2)**0.5
    BC = ((x3-x2)**2 + (y3-y2)**2)**0.5
    n = math.acos(A/(BA*BC))
    return math.degrees(n)
x1, y1, x2, y2, x3, y3 = map(int, input().split())
print(angle(x1, y1, x2, y2, x3, y3))
                  
```

W5A30: Viết hàm tính hệ số nhị thức C(n, k).

```python
def hesonhithuc(n, k):
    if k <= n:
        tich1, tich2, tich3 = 1, 1, 1
        for i in range(1, n+1): #n!
            tich1 *= i
        for j in range(1, k+1): #k!
            tich2 *= j
        for o in range(1, n-k+1): #(n-k)!
            tich3 *= o
        return int(tich1/(tich2*tich3))
    else:
        return f'Hệ số không thỏa mãn'
n, k = map(int, input().split())
print(hesonhithuc(n, k))
        
```

W5A31: Viết hàm tính tổng bình phương các số từ 1 đến n.

```python
def tongbphuong(n):
    tong = 0
    for i in range(1, n+1):
        tong += i**2
    return tong
n = int(input())
print(tongbphuong(n))
```
