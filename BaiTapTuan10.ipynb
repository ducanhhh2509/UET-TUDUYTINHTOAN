## W10A1: [DivideByZero] Nhập vào hai số nguyên a, b. Thực hiện in ra kết quả phép chia a/b. Bắt lỗi ngoại lệ khi b bằng 0. Kết quả in ra chính xác tới 2 chữ số thập phân nếu b khác 0. (Dùng try ... except bắt lỗi ZeroDivisionError) 

```python
try:
    a, b = map(int, input().split())
    n = a/b
    print("%.2f"%n)
except ZeroDivisionError:
    print(f"Division by zero")
except ValueError:
    print(f"Value Error")
```

## W10A3: [ValueError - HappyNumbers]
## Nhập vào một số nguyên dương a, Kiểm tra xem a có phải là số hạnh phúc không. Một số nguyên dương n được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước sau:
## - Tính tổng bình phương các chữ số của n.
## - Lặp lại bước đó với kết quả vừa tính.
## - Nếu cuối cùng kết quả là 1, thì n là số hạnh phúc.
## - Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như 4 → 16 → 37 → ...), thì n không phải là số hạnh phúc.
## Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try ... except bắt lỗi ValueError thêm set) 

```python
def happy(n):
    tong = 0
    if n == 1:
        return f"happy"
    elif n == 4:
        return f"no happy"
    while n > 0:
        last_digit = n % 10
        tong += last_digit**2
        n //= 10
    return happy(tong)
try:
    n = int(input())
    print(happy(n))
except ValueError:
    print(f"ValueError")   
```

## W10A4: Sử dụng try except,kết hợp cùng else... finally. Tính tổng 2 số, nếu nhập sai kiểu thì thông báo, luôn luôn in "Kết thúc chương trình." sau khi chạy.

```python
try:
    a, b = map(int, input().split())
except ValueError:
    print(f"ValueError")
else:
    print(a + b)
finally:
    print(f"Kết thúc chương trình")
```

## W10A14: [TypeError - SumNumbers] Nhập vào một danh sách gồm nhiều giá trị. Hãy cộng tất cả phần tử sau khi chuyển được sang int. Nếu phần tử nào không chuyển được thì bỏ qua. In ra tổng cuối cùng.

```python
arr = list(map(str, input().split()))
tong = 0
for i in arr:
    try:
        tong += int(i)
    except TypeError, ValueError:
        pass
else:
    print(tong)
```

## W10A15: [Raise_Exception - PositiveNumber] Nhập vào một số nguyên n. Nếu n <= 0 thì chủ động phát sinh ngoại lệ và in ra "So phai duong". Nếu hợp lệ thì in ra bình phương của n. 

```python
try:
    n = int(input())
    if n <= 0:
        raise ValueError(f"So phai duong")
except ValueError as e:
    if str(e) == f"So phai duong":
        print(f"So phai duong")
    else:
        print(f"Invalid")
```
